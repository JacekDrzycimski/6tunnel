```python
# Newton's Law of Universal Gravitation
# F = G * (m1 * m2) / r^2

# Gravitational constant in m^3 kg^-1 s^-2
G = 6.67430e-11

def gravitational_force(m1, m2, r):
    """
    Calculates the gravitational force between two objects.
    m1: Mass of the first object (kg)
    m2: Mass of the second object (kg)
    r: Distance between the centers of the masses (m)
    """
    if r <= 0:
        raise ValueError("Distance r must be greater than 0")
    return G * (m1 * m2) / (r ** 2)

# Example: Force between Earth (5.972e24 kg) and an object of 70 kg on its surface (radius ~6.371e6 m)
m_earth = 5.972e24
m_object = 70
r_earth = 6.371e6

force = gravitational_force(m_earth, m_object, r_earth)
print(f"Gravitational force: {force:.2f} N")
```

```python
import math

def escape_velocity(M, R):
    """
    Calculates the escape velocity from a celestial body.
    M: Mass of the body (kg)
    R: Radius of the body (m)
    """
    if R <= 0:
        raise ValueError("Radius R must be greater than 0")
    return math.sqrt(2 * G * M / R)

# Calculate escape velocity for Earth
v_esc = escape_velocity(m_earth, r_earth)

print(f"Escape velocity from Earth: {v_esc:.2f} m/s")
print(f"Escape velocity from Earth: {v_esc / 1000:.2f} km/s")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define distances starting from Earth's surface out to 5 times the Earth's radius
distances = np.linspace(r_earth, r_earth * 5, 500)

# Calculate gravitational force for each distance
forces = [gravitational_force(m_earth, m_object, r) for r in distances]

# Convert distances to thousands of kilometers for easier reading on the plot
distances_km = distances / 1e3

# Plotting
fig = plt.figure(figsize=(10, 6))
plt.plot(distances_km, forces, label='Gravitational Force (N)', color='royalblue', linewidth=2)
plt.axvline(x=r_earth/1e3, color='crimson', linestyle='--', label="Earth's Surface")

plt.title('Gravitational Force on a 70kg Object vs. Distance from Earth Center')
plt.xlabel('Distance from Center (km)')
plt.ylabel('Force (N)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
def gravitational_potential_energy(m1, m2, r):
    """
    Calculates the gravitational potential energy between two objects.
    m1: Mass of the first object (kg)
    m2: Mass of the second object (kg)
    r: Distance between the centers of the masses (m)
    """
    if r <= 0:
        raise ValueError("Distance r must be greater than 0")
    return -G * (m1 * m2) / r

# Calculate gravitational potential energy for each distance
gpe = [gravitational_potential_energy(m_earth, m_object, r) for r in distances]

# Plotting GPE vs. Distance
fig2 = plt.figure(figsize=(10, 6))
plt.plot(distances_km, gpe, label='Gravitational Potential Energy (J)', color='darkorange', linewidth=2)
plt.axvline(x=r_earth/1e3, color='crimson', linestyle='--', label="Earth's Surface")

plt.title('Gravitational Potential Energy of a 70kg Object vs. Distance from Earth Center')
plt.xlabel('Distance from Center (km)')
plt.ylabel('Potential Energy (Joules)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# Display values at Earth's surface and at 5x Earth's radius
print(f"GPE at Earth's surface: {gpe[0]:.2e} J")
print(f"GPE at 5x Earth's radius: {gpe[-1]:.2e} J")
```
