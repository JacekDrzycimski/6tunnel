### How to Change Username under Ubuntu Linux

To change a username (e.g., from `old_username` to `new_username`), follow these standard steps.

> **Important:** It is highly recommended to do this while logged in as a different administrator account (not the one being modified) or in Recovery Mode. You cannot change the username of a user that is currently logged in.

#### Step 1: Log out of the target account
Ensure no processes are running under the old user:
```bash
sudo killall -u old_username
```

#### Step 2: Change the username
Use the `usermod` command to change the login name:
```bash
sudo usermod -l new_username old_username
```

#### Step 3: Change the home directory (Optional but Recommended)
To match the home directory name with the new username and move all files:
```bash
sudo usermod -d /home/new_username -m new_username
```

#### Step 4: Change the user's group name
Usually, primary groups share the same name as the user. To update the primary group name:
```bash
sudo groupmod -n new_username old_username
```

````python
# We can use Python to print a reminder of the syntax to copy/paste directly
old_user = "old_username"
new_user = "new_username"

print(f"# Run these commands in your Ubuntu terminal:")
print(f"sudo killall -u {old_user}")
print(f"sudo usermod -l {new_user} {old_user}")
print(f"sudo usermod -d /home/{new_user} -m {new_user}")
print(f"sudo groupmod -n {new_user} {old_user}")
````
