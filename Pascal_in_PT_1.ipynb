# Running Pascal Code in Python

To execute Pascal code from Python, we can:
1. Install the Free Pascal Compiler (`fpc`).
2. Write the Pascal source code to a `.pas` file.
3. Compile the file using `fpc` via Python's `subprocess` module.
4. Run the generated executable and capture its output.

```python
# 1. Install Free Pascal Compiler
!apt-get install -y fpc
```

```python
import subprocess

# 2. Define a simple Pascal program
pascal_code = """
program HelloWorld;
begin
  writeln('Hello from Pascal code executed inside Python!');
end.
"""

# Save the Pascal code to a file
with open("hello.pas", "w") as f:
    f.write(pascal_code)

print("Pascal file written successfully.")
```

```python
# 3. Compile the Pascal file
compile_process = subprocess.run(["fpc", "hello.pas"], capture_output=True, text=True)

if compile_process.returncode == 0:
    print("Compilation successful! Running the executable...\n")
    # 4. Run the compiled executable
    run_process = subprocess.run(["./hello"], capture_output=True, text=True)
    print("--- Pascal Output ---")
    print(run_process.stdout)
else:
    print("Compilation failed:")
    print(compile_process.stderr)
```

```python
import subprocess

# Compile hello.pas using the subprocess module
compile_command = ["fpc", "hello.pas"]
result = subprocess.run(compile_command, capture_output=True, text=True)

# Display compilation results
print("STDOUT:")
print(result.stdout)
print("STDERR:")
print(result.stderr)
print(f"Exit Code: {result.returncode}")
```

## Performing Calculations with Pascal inside Python

We will now write a new Pascal program (`calc.pas`) that defines some variables, performs arithmetic operations (addition, subtraction, multiplication), and prints the results.

```python
# 1. Write the Pascal calculation code
calc_pascal_code = """
program Calculator;
var
  num1, num2: Integer;
  sum, diff, prod: Integer;
begin
  num1 := 15;
  num2 := 4;

  sum := num1 + num2;
  diff := num1 - num2;
  prod := num1 * num2;

  writeln('--- Pascal Calculator ---');
  writeln('Number 1: ', num1);
  writeln('Number 2: ', num2);
  writeln('Sum: ', sum);
  writeln('Difference: ', diff);
  writeln('Product: ', prod);
end.
"""

with open("calc.pas", "w") as f:
    f.write(calc_pascal_code)

print("calc.pas written successfully.")
```

```python
# 2. Compile and run the calculation program
import subprocess

# Compile
compile_res = subprocess.run(["fpc", "calc.pas"], capture_output=True, text=True)

if compile_res.returncode == 0:
    print("Compilation Successful! Running program:\n")
    # Run executable
    run_res = subprocess.run(["./calc"], capture_output=True, text=True)
    print(run_res.stdout)
else:
    print("Compilation Failed!")
    print(compile_res.stderr)
```

## Loops and Conditions in Pascal

Now, let's create a Pascal program (`loops_and_conds.pas`) that uses a `for` loop to iterate from 1 to 10 and an `if-then-else` conditional statement to classify each number as even or odd.

```python
# 1. Write the Pascal source code with a loop and conditional statements
loops_pascal_code = """
program LoopsAndConditions;
var
  i: Integer;
begin
  writeln('--- Classification of Numbers (1 to 10) ---');
  for i := 1 to 10 do
  begin
    write('Number ', i, ' is ');
    if (i mod 2 = 0) then
      writeln('Even')
    else
      writeln('Odd');
  end;
end.
"""

with open("loops_and_conds.pas", "w") as f:
    f.write(loops_pascal_code)

print("loops_and_conds.pas written successfully.")
```

```python
# 2. Compile and run the loops and conditions program
import subprocess

compile_res = subprocess.run(["fpc", "loops_and_conds.pas"], capture_output=True, text=True)

if compile_res.returncode == 0:
    print("Compilation Successful! Running program:\n")
    run_res = subprocess.run(["./loops_and_conds"], capture_output=True, text=True)
    print(run_res.stdout)
else:
    print("Compilation Failed!")
    print(compile_res.stderr)
```
