### 1. Sending Email using SMTP (Gmail Example)
To send emails via Gmail, you will typically need to generate an **App Password** from your Google Account settings if you have 2-Factor Authentication enabled.

```python
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText

def send_email(sender_email, sender_password, receiver_email, subject, body):
    # Set up the SMTP server (using Gmail SMTP in this example)
    smtp_server = "smtp.gmail.com"
    smtp_port = 587

    # Create message container
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = receiver_email
    msg['Subject'] = subject

    # Attach the body of the message
    msg.attach(MIMEText(body, 'plain'))

    try:
        # Connect to server and send email
        server = smtplib.SMTP(smtp_server, smtp_port)
        server.starttls() # Secure the connection
        server.login(sender_email, sender_password)
        server.sendmail(sender_email, receiver_email, msg.as_string())
        server.quit()
        print("Email sent successfully!")
    except Exception as e:
        print(f"An error occurred while sending email: {e}")

# --- Demo / Configuration ---
# SENDER_EMAIL = 'your_email@gmail.com'
# SENDER_PASSWORD = 'your_app_password_here'
# RECEIVER_EMAIL = 'target_email@example.com'
# send_email(SENDER_EMAIL, SENDER_PASSWORD, RECEIVER_EMAIL, 'Test Subject', 'Hello from Colab!')
```

```python
from twilio.rest import Client

def send_sms(account_sid, auth_token, from_number, to_number, message_body):
    try:
        client = Client(account_sid, auth_token)
        message = client.messages.create(
            body=message_body,
            from_=from_number,
            to=to_number
        )
        print(f"SMS sent successfully! Message SID: {message.sid}")
    except Exception as e:
        print(f"An error occurred while sending SMS: {e}")

# --- Demo / Configuration ---
# ACCOUNT_SID = 'your_twilio_account_sid'
# AUTH_TOKEN = 'your_twilio_auth_token'
# FROM_NUMBER = '+1234567890'  # Your Twilio phone number
# TO_NUMBER = '+19876543210'    # Recipient phone number
# send_sms(ACCOUNT_SID, AUTH_TOKEN, FROM_NUMBER, TO_NUMBER, 'Hello! This is a test text message.')
```

### General Secure Storage of API Credentials in Colab

To securely store any API credentials or sensitive information in your Colab notebook, you should use the **Secrets Manager**.

1.  **Access the Secrets Manager**: Click on the '🔑' icon (Secrets) in the left-hand panel of your Colab notebook.
2.  **Add New Secret**: Click 'Add new secret'.
3.  **Define Name and Value**: Enter a descriptive name for your secret (e.g., `MY_API_KEY`, `DATABASE_PASSWORD`) and paste the actual credential as its value.
4.  **Enable Notebook Access**: Crucially, toggle the 'Notebook access' switch **ON** for each secret you want your current notebook to use. This makes the secret accessible via `google.colab.userdata`.

Once configured, you can retrieve these secrets in your Python code as demonstrated below. This keeps your credentials out of your notebook's code history and makes them easily reusable across different notebooks without hardcoding them.

```python
from google.colab import userdata

# To demonstrate, let's assume you've added a secret named 'MY_GENERIC_API_KEY'
# to the Secrets Manager and enabled 'Notebook access' for it.

try:
    # Retrieve your generic API key from Colab's Secret Manager
    my_api_key = userdata.get('MY_GENERIC_API_KEY')
    print(f"Successfully retrieved secret value: {my_api_key[:5]}... (first 5 chars shown)") # Print partially for security
    # You can now use 'my_api_key' in your code, e.g., for authenticating API calls
except userdata.SecretNotFoundError:
    print("Error: Secret 'MY_GENERIC_API_KEY' not found.")
    print("Please add it to Colab Secrets (🔑 icon) and ensure 'Notebook access' is ON.")
except userdata.NotebookAccessError:
    print("Error: Secret 'MY_GENERIC_API_KEY' exists, but 'Notebook access' is not enabled.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
