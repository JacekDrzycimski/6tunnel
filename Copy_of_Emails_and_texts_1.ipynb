### 1. Sending Email using SMTP (Gmail Example)
To send emails via Gmail, you will typically need to generate an **App Password** from your Google Account settings if you have 2-Factor Authentication enabled.

```python
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText

def send_email(sender_email, sender_password, receiver_email, subject, body):
    # Set up the SMTP server (using Gmail SMTP in this example)
    smtp_server = "smtp.gmail.com"
    smtp_port = 587

    # Create message container
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = receiver_email
    msg['Subject'] = subject

    # Attach the body of the message
    msg.attach(MIMEText(body, 'plain'))

    try:
        # Connect to server and send email
        server = smtplib.SMTP(smtp_server, smtp_port)
        server.starttls() # Secure the connection
        server.login(sender_email, sender_password)
        server.sendmail(sender_email, receiver_email, msg.as_string())
        server.quit()
        print("Email sent successfully!")
    except Exception as e:
        print(f"An error occurred while sending email: {e}")

# --- Demo / Configuration ---
# SENDER_EMAIL = 'your_email@gmail.com'
# SENDER_PASSWORD = 'your_app_password_here'
# RECEIVER_EMAIL = 'target_email@example.com'
# send_email(SENDER_EMAIL, SENDER_PASSWORD, RECEIVER_EMAIL, 'Test Subject', 'Hello from Colab!')
```

### 2. Sending SMS via Twilio API
To send SMS messages, you can install and use the `twilio` library. You will need a Twilio Account SID, Auth Token, and a Twilio phone number.

```python
# Install Twilio SDK
!pip install twilio
```

```python
from twilio.rest import Client

def send_sms(account_sid, auth_token, from_number, to_number, message_body):
    try:
        client = Client(account_sid, auth_token)
        message = client.messages.create(
            body=message_body,
            from_=from_number,
            to=to_number
        )
        print(f"SMS sent successfully! Message SID: {message.sid}")
    except Exception as e:
        print(f"An error occurred while sending SMS: {e}")

# --- Demo / Configuration ---
# ACCOUNT_SID = 'your_twilio_account_sid'
# AUTH_TOKEN = 'your_twilio_auth_token'
# FROM_NUMBER = '+1234567890'  # Your Twilio phone number
# TO_NUMBER = '+19876543210'    # Recipient phone number
# send_sms(ACCOUNT_SID, AUTH_TOKEN, FROM_NUMBER, TO_NUMBER, 'Hello! This is a test text message.')
```

### Securely Storing Twilio Credentials in Colab

1.  **Open the Secrets Manager**: Click on the '🔑' icon (Secrets) in the left-hand panel of your Colab notebook.
2.  **Add Secrets**: Click 'Add new secret'.
    *   For your Account SID, enter `TWILIO_ACCOUNT_SID` as the name and paste your Account SID as the value.
    *   For your Auth Token, enter `TWILIO_AUTH_TOKEN` as the name and paste your Auth Token as the value.
    *   For your Twilio Phone Number, enter `TWILIO_FROM_NUMBER` as the name and paste your Twilio phone number (e.g., `+1234567890`) as the value.
3.  **Enable Notebook Access**: Make sure the 'Notebook access' toggle is enabled for each secret you add. This allows your notebook code to access these secrets.

Once configured, you can retrieve these secrets in your Python code using `userdata.get()`:

```python
from google.colab import userdata

# Retrieve secrets from Colab's Secret Manager
# Make sure to add TWILIO_ACCOUNT_SID, TWILIO_AUTH_TOKEN, and TWILIO_FROM_NUMBER
# to the Secrets (🔑) panel in the left sidebar and toggle "Notebook access" to ON.
try:
    ACCOUNT_SID = userdata.get('TWILIO_ACCOUNT_SID')
    AUTH_TOKEN = userdata.get('TWILIO_AUTH_TOKEN')
    FROM_NUMBER = userdata.get('TWILIO_FROM_NUMBER')

    # Define the recipient number and message body
    TO_NUMBER = '+19876543210' # Replace with the recipient's phone number
    message_body = 'Hello from Colab! This is a secure test text message using Twilio.'

    # Call the send_sms function with the retrieved credentials
    send_sms(ACCOUNT_SID, AUTH_TOKEN, FROM_NUMBER, TO_NUMBER, message_body)
except userdata.SecretNotFoundError as e:
    print(f"Configuration Error: {e}\n")
    print("Please click on the Key icon (🔑) in the left sidebar and add the missing secret with the exact name and turn on 'Notebook access'.")
```
